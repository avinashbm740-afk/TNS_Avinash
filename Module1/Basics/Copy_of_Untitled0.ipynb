### Simple Command-Line Greeting Application

This is a basic Python script that acts as an interactive command-line application. It takes user input and provides a simple greeting.

```python
def Greeting():
    print("Welcome to the Simple Greeting App!")
    print("---------------------------------")

    name = input("Please enter your name: ")

    if name:
        print(f"Hello, {name}! It's great to meet you.")
    else:
        print("Hello there! I guess you prefer to remain anonymous.")

    print("---------------------------------")
    print("Thank you for using the app!")

Greeting()
```

#### 2. Methods within a Class

Methods are functions that belong to a class. They operate on the data (attributes) of an object created from that class. You call them using the object instance.

#### 2. Listing methods available for a specific object or type

You can use the `dir()` function on an object or a type to see all its attributes, including its methods. Methods are usually functions that belong to an object.

```python
print("--- Methods for a String Object ---")
my_string = "hello world"
string_methods = [attr for attr in dir(my_string) if callable(getattr(my_string, attr)) and not attr.startswith('__')]
print(f"Number of string methods: {len(string_methods)}")
print("First 10 string methods (alphabetical):", sorted(string_methods)[:10])

print("\n--- Methods for a List Object ---")
my_list = [1, 2, 3]
list_methods = [attr for attr in dir(my_list) if callable(getattr(my_list, attr)) and not attr.startswith('__')]
print(f"Number of list methods: {len(list_methods)}")
print("List methods (alphabetical):", sorted(list_methods))

print("\n--- Methods for a Dictionary Object ---")
my_dict = {'a': 1, 'b': 2}
dict_methods = [attr for attr in dir(my_dict) if callable(getattr(my_dict, attr)) and not attr.startswith('__')]
print(f"Number of dictionary methods: {len(dict_methods)}")
print("Dictionary methods (alphabetical):", sorted(dict_methods))
```

### Storing Fixed Latitude and Longitude Coordinates

To ensure that latitude and longitude values for a college are fixed and cannot be changed after creation, we can use a Python class with properties that only have getters. This makes the attributes effectively read-only.

```python
class CollegeLocation:
    """Represents the fixed latitude and longitude of a college.

    The latitude and longitude are immutable once set during initialization.
    """

    def __init__(self, latitude, longitude):
        # Store the values internally
        self._latitude = latitude
        self._longitude = longitude

    @property
    def latitude(self):
        """Get the fixed latitude."""
        return self._latitude

    @property
    def longitude(self):
        """Get the fixed longitude."""
        return self._longitude

    def __str__(self):
        return f"College Location: Latitude={self.latitude}, Longitude={self.longitude}"

    def __repr__(self):
        return f"CollegeLocation(latitude={self.latitude}, longitude={self.longitude})"

# Create an instance of CollegeLocation with fixed coordinates
# Example: Googleplex coordinates
alma_mater_coords = CollegeLocation(latitude=37.422131, longitude=-122.084478)

print(alma_mater_coords)
print(f"Latitude: {alma_mater_coords.latitude}")
print(f"Longitude: {alma_mater_coords.longitude}")

# Attempting to change the values will raise an AttributeError
try:
    alma_mater_coords.latitude = 34.052235  # This line will cause an error
except AttributeError as e:
    print(f"\nAttempted to change latitude, caught expected error: {e}")

try:
    alma_mater_coords.longitude = -118.243683 # This line will cause an error
except AttributeError as e:
    print(f"Attempted to change longitude, caught expected error: {e}")
```

### Storing Fixed Coordinates Using Tuples

Tuples are an excellent choice for storing fixed data like geographical coordinates because they are **immutable**. Once a tuple is created, its elements cannot be changed, ensuring the values remain constant.

```python
# Define the college's fixed latitude and longitude using a tuple
# Format: (latitude, longitude)
college_data = (37.422131, -122.084478)

print(f"College Coordinates (Tuple): {college_data}")
print(f"Latitude: {college_data[0]}")
print(f"Longitude: {college_data[1]}")

# Attempting to modify a tuple element will result in a TypeError
try:
    college_data[0] = 34.052235
except TypeError as e:
    print(f"\nerror: {e}")
    print("This demonstrates the immutability of tuples.\n")

# You can also assign the tuple elements to named variables for clarity
latitude, longitude = college_data
print(f"Named variables: Latitude={latitude}, Longitude={longitude}")
```
